```python
import pandas as pd
import random
from faker import Faker

random.seed(42)
Faker.seed(42)

fake = Faker("en_IN")

departments = [
    "Cardiology", "Neurology", "Orthopedics", "Pediatrics",
    "General Medicine", "Gynecology", "Oncology", "ENT"
]

doctors = [
    "Dr. Sharma", "Dr. Singh", "Dr. Gupta", "Dr. Verma",
    "Dr. Patel", "Dr. Khan", "Dr. Roy", "Dr. Das"
]

diseases = [
    "Diabetes", "Hypertension", "Asthma", "Fever",
    "Fracture", "COVID-19", "Heart Disease", "Cancer",
    "Migraine", "Kidney Stone"
]

treatments = [
    "Medication", "Surgery", "Therapy",
    "Chemotherapy", "Observation", "Physiotherapy"
]

cities = [
    "Patna", "Delhi", "Mumbai", "Kolkata",
    "Lucknow", "Ranchi", "Bhopal", "Jaipur"
]

records = []

for i in range(1, 100001):

    admission = fake.date_between(start_date="-2y", end_date="today")
    stay = random.randint(1, 15)
    discharge = admission + pd.Timedelta(days=stay)

    bill = random.randint(5000, 200000)

    records.append({
        "Patient_ID": f"P{i:05d}",
        "Patient_Name": fake.unique.name(),
        "Age": random.randint(1, 90),
        "Gender": random.choice(["Male", "Female"]),
        "City": random.choice(cities),
        "Admission_Date": admission,
        "Discharge_Date": discharge,
        "Department": random.choice(departments),
        "Doctor_Name": random.choice(doctors),
        "Disease": random.choice(diseases),
        "Treatment": random.choice(treatments),
        "Bed_ID": f"B{random.randint(1,300):03d}",
        "Admission_Type": random.choice(["Emergency","OPD","Referral"]),
        "Status": random.choice(["Admitted","Discharged"]),
        "Bill_Amount": bill,
        "Outcome": random.choice(["Recovered","Under Treatment","Transferred"])
    })

df = pd.DataFrame(records)

df.to_csv("medical_operations_100000.csv", index=False)

#print("Dataset created successfully!")
print(df.tail())
```
